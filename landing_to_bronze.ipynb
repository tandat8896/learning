```python
%sql
create schema if not exists etl_projects.bronze
```

```python
landing_path = "/Volumes/etl_projects/landing/thpc"
tables = [
    x.name.rstrip("/")
    for x in dbutils.fs.ls(landing_path)
]
print(tables)

for table in tables:
    df = spark.read.parquet(f"{landing_path}/{table}", header=True, inferSchema=True)
    (
        df.write
        .format("delta")
        .mode("overwrite")
        .saveAsTable(f"etl_projects.bronze.{table}")
    )
```
