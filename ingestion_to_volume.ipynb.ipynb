```python
import pyspark
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("ingestion").getOrCreate()
```

```python
%%sql
-- create catalog if not exists etl_projects;

create schema if not exists etl_projects.landing;

create volume if not exists etl_projects.landing.thpc;

```

```python
%%sql
show catalogs;
show schemas in etl_projects;
show volumes in etl_projects.landing
```

```python
%%sh
ls -la /Volumes/etl_projects/landing/thpc
```

```python
tables = [
    "customer",
    "lineitem",
    "nation",
    "orders",
    "part",
    "partsupp",
    "region",
    "supplier"
]
landing_path = "/Volumes/etl_projects/landing/thpc"

for table_name in tables:
    df = spark.table(f"samples.tpch.{table_name}")
    df.write.mode("overwrite").format("parquet").save(f"{landing_path}/{table_name}")
    print(f"{table_name} written to {landing_path}/{table_name}")
```
